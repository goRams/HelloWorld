```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Parameters
fs = 1000 # Sampling frequency
f = 5 # Input signal frequency
t = np.arange(0, 1, 1/fs) # Time vector

# Input signal
x = np.sin(2 * np.pi * f * t)

# Sigma-delta modulator
y = np.zeros_like(x)
integrator = 0
for i in range(len(x)):
    integrator += x[i] - y[i-1]
    y[i] = 1 if integrator > 0 else -1

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(t, x, label='Input Signal')
plt.step(t, y, label='Sigma-Delta Output', where='post')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.legend()
plt.title('First-Order Sigma-Delta Modulator')
plt.grid()
plt.show()

fig = go.Figure()
fig.add_trace(go.Scatter(x=t, y=x, mode='lines', name='Input Signal'))
fig.add_trace(go.Scatter(x=t, y=y, mode='lines', name='Sigma-Delta Output'))
fig.update_layout(title='First-Order Sigma-Delta Modulator',
                  xaxis_title='Time [s]',
                  yaxis_title='Amplitude')
fig.show()
```
